```python
# Function to encode the image
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')
```

```python
def genfeatures(base64_image):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
        {
        "role": "user",
        "content": [
            {
            "type": "text",
            "text": "Describe the main features of a clothing item using single words or phrases for each category: 'product name,' 'product type,' 'pattern,' 'color,' 'color shade,' 'material,' ,'occasion' and 'applicable season.' Additionally, provide a single descriptive sentence on item's  other notable details.",
            },
            {
            "type": "image_url",
            "image_url": {
            "url":  f"data:image/jpeg;base64,{base64_image}"
            },
            },
        ],
    }
    ],
    )
    print(response.choices[0])
    return response.choices[0]
```
