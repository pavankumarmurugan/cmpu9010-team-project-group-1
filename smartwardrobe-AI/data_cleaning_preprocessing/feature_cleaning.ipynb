```python
def dig_des(des):
    try:
        match = re.search(r"content='(.*?)'", des)
        content = match.group(1)
        content = content.replace("\\n", "\n")
    except:
        match = content_match = re.search(r"content=\"(.*?)\", refusal=", des, re.DOTALL)
        content = match.group(1)
        content = content.replace("\\n", "\n")
    return content
    
    
```

```python
def extract_features(description):
    #content = dict(re.findall(r'- \*\*(.*?)\*\*:\s*(.*)', description))
    features = {
        'Product_name': re.search(r'- \*\*product name:\*\* (.*)', description),
        'Product_type': re.search(r'- \*\*product type:\*\* (.*)', description),
        'Pattern': re.search(r'- \*\*pattern:\*\* (.*)', description),
        'Color': re.search(r'- \*\*color:\*\* (.*)', description),
        'Color_shade': re.search(r'- \*\*color shade:\*\* (.*)', description),
        'Material': re.search(r'- \*\*material:\*\* (.*)', description),
        'Occasion': re.search(r'- \*\*occasion:\*\* (.*)', description),
        'Applicable_season': re.search(r'- \*\*applicable season:\*\* (.*)', description),
        #'Description': description.split("\n\n")[-1].strip()
    }
    # Extract groups and handle any missing matches
    return {key: (match.group(1).strip() if match else None) for key, match in features.items()}

# Apply extraction function to the DataFrame
df_features = ai_data_2['Description_1'].apply(extract_features).apply(pd.Series)
```

```python
def dess(des1):
    des1 = des1.replace("**notable details:**"," ")
    des1 = des1.replace("**other notable details:**"," ")
    des1 = des1.replace("**notable details**:"," ")
    des1 = des1.replace("**other notable details**:"," ")
    des1 = des1.replace("notable details:"," ")
    des1 = des1.replace('notable detail:'," ")
    des1 = des1.split("\n\n")[-1].strip()
    return des1
```
