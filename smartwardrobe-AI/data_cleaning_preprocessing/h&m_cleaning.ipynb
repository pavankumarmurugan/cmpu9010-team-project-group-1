```python
stop_words = set(stopwords.words('english'))
stop_words = list(stop_words)
```

```python
def clean_text(list1):
    def lower1(text):
        return text.lower()
    def remove_noise(text):   
        #remove lowercase
        text = text.apply(lambda x:" ".join(x.lower() for x in x.split()))
        
        #remove whitespace
        text = text.apply(lambda x:" ".join(x.strip() for x in x.split()))
        
        #remove special characters
        text = text.apply(lambda x:"".join([" " if ord(i) <32 or ord(i) >126 else i for i in x]))
        
        #remove punctuation
        text = text.str.replace('[^\w\s]','',regex=True)
        
        #remove numbers
        text = text.str.replace('\d','',regex=True)
        
        #remove stopwords
        text = text.apply(lambda x:" ".join([word for word in x.split() if word not in list(stop_words)]))
        
        return text
    list1[['description']] = list1[['description']].applymap(lower1).copy()
    list1[['description']].reset_index(drop=True).copy()
    list1['cleaned'] = remove_noise(list1['description'])
    return list1['cleaned']
    
```

```python

# Load BERT tokenizer and model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=4)  # Adjust num_labels according to your classes
```

```python

# Function to make predictions using batching
def predict(model, texts, batch_size=32):
    model.eval()  # Set the model to evaluation mode
    predictions = []

    # Process texts in batches
    for i in range(0, len(texts), batch_size):
        batch_texts = texts[i:i + batch_size]

        # Tokenize the batch
        batch_encodings = tokenizer(batch_texts, 
                                     add_special_tokens=True,
                                     max_length=128,
                                     padding=True,
                                     truncation=True,
                                     return_tensors='pt')

        input_ids = batch_encodings['input_ids']
        attention_mask = batch_encodings['attention_mask']

        with torch.no_grad():
            # Make prediction
            outputs = model(input_ids=input_ids.cuda(), attention_mask=attention_mask.cuda())
            _, preds = torch.max(outputs.logits, dim=1)
            predictions.extend(preds.cpu().numpy())  # Append the predicted classes

    return predictions
```

```python
from transformers import AutoTokenizer, AutoModel
import torch
import torch.nn.functional as F
```

```python

#Mean Pooling - Take attention mask into account for correct averaging
def mean_pooling(model_output, attention_mask):
    token_embeddings = model_output[0] #First element of model_output contains all token embeddings
    input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
    return torch.sum(token_embeddings * input_mask_expanded, 1) / torch.clamp(input_mask_expanded.sum(1), min=1e-9)
```

```python

# Load model from HuggingFace Hub
tokenizer = AutoTokenizer.from_pretrained('sentence-transformers/all-MiniLM-L6-v2')
model = AutoModel.from_pretrained('sentence-transformers/all-MiniLM-L6-v2')
```

```python

# Tokenize sentences
encoded_input = tokenizer(list(X), padding=True, truncation=True, return_tensors='pt')
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
model.to(device)
```

```python

# Compute token embeddings
with torch.no_grad():
    model_output = model(**encoded_input)
```

```python

# Perform pooling
sentence_embeddings = mean_pooling(model_output, encoded_input['attention_mask'])
```

```python

# Normalize embeddings
sentence_embeddings = F.normalize(sentence_embeddings, p=2, dim=1)
```
